```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', 100)

transaction = pd.read_csv('../data/raw/train_transaction.csv')
identity = pd.read_csv('../data/raw/train_identity.csv')

print(f"Transaction shape: {transaction.shape}")
print(f"Identity shape: {identity.shape}")
```

```python
df = transaction.merge(identity, on='TransactionID', how='left')
print(f"Merged shape: {df.shape}")
df.head()
```

```python
fraud_rate = df['isFraud'].mean()
print(f"Fraud rate: {fraud_rate:.4%}")

sns.countplot(x='isFraud', data=df)
plt.title('Class Distribution: Fraud vs Legit')
plt.show()
```

```python
missing = df.isnull().mean().sort_values(ascending=False)
missing[missing > 0].head(30)
```

```python
edge_candidate_cols = ['card1', 'card2', 'card3', 'card4', 'card5', 'card6',
                        'addr1', 'addr2', 'P_emaildomain', 'R_emaildomain',
                        'DeviceType', 'DeviceInfo']
df[edge_candidate_cols].nunique()
```

```python
card_fraud = df.groupby('card1')['isFraud'].agg(['mean', 'count'])
card_fraud_shared = card_fraud[card_fraud['count'] > 1]
print(f"Cards used in >1 transaction: {len(card_fraud_shared)}")
print(f"Fraud rate among shared cards: {card_fraud_shared['mean'].mean():.4%}")
print(f"Fraud rate overall: {fraud_rate:.4%}")
```

```python
print(f"Memory usage: {df.memory_usage(deep=True).sum() / 1e9:.2f} GB")
```

```python
for col in ['card1', 'card2', 'addr1', 'P_emaildomain', 'DeviceInfo']:
    grp = df.groupby(col)['isFraud'].agg(['mean', 'count'])
    shared = grp[grp['count'] > 1]
    print(f"{col}: shared-entity fraud rate = {shared['mean'].mean():.4%}, "
          f"n_shared_entities = {len(shared)}")
```

```python
for col in ['card1', 'card2', 'addr1', 'P_emaildomain', 'DeviceInfo']:
    grp = df.groupby(col)['isFraud'].agg(['mean', 'count'])
    shared = grp[grp['count'] > 1]
    print(f"{col}: shared-entity fraud rate = {shared['mean'].mean():.4%}, "
          f"n_shared_entities = {len(shared)}")
```

```python
df['combo_key'] = (df['card1'].astype(str) + '_' +
                    df['addr1'].astype(str) + '_' +
                    df['P_emaildomain'].astype(str))

combo_grp = df.groupby('combo_key')['isFraud'].agg(['mean', 'count'])
combo_shared = combo_grp[combo_grp['count'] > 1]
print(f"Shared combo-key fraud rate: {combo_shared['mean'].mean():.4%}")
print(f"n_shared combo entities: {len(combo_shared)}")
```

```python
for col in ['card1', 'addr1', 'P_emaildomain', 'DeviceInfo']:
    grp = df.groupby(col)['isFraud'].agg(['mean', 'count'])
    shared = grp[grp['count'] > 1]
    shared_txns = df[df[col].isin(shared.index)]
    weighted_rate = shared_txns['isFraud'].mean()
    print(f"{col}: transaction-weighted shared fraud rate = {weighted_rate:.4%}, "
          f"n_transactions = {len(shared_txns)}")
```
